## 2. Descriptivos

La tabla numérica debe tener `n`, media, desviación estándar, mínimo, p25, mediana, p75 y máximo. Para las dummies, usa frecuencias y porcentajes.

**Pista:** `mean()`, `sd()`, `quantile()`, `median()` y `table()` son suficientes. Compara media con mediana para discutir asimetría y desviación estándar con los percentiles para discutir dispersión.

```r
variables_numericas <- c("wage", "lwage", "educ", "exper", "tenure")

describir <- function(nombre) {
  x <- wage_df[[nombre]]
  data.frame(variable = nombre, n = sum(!is.na(x)), media = mean(x), sd = sd(x),
             minimo = min(x), p25 = quantile(x, .25), mediana = median(x),
             p75 = quantile(x, .75), maximo = max(x))
}
do.call(rbind, lapply(variables_numericas, describir))

variables_binarias <- c("female", "married", "nonwhite")
lapply(variables_binarias, function(nombre) {
  f <- table(wage_df[[nombre]], useNA = "ifany")
  data.frame(variable = nombre, valor = names(f), frecuencia = as.vector(f),
             porcentaje = 100 * as.vector(f) / sum(f))
})
```
